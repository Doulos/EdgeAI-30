### **Step 2.** Use of Cell Magics Commands
Colab cells can execute bash shell commands using [Cell Magics](https://ipython.readthedocs.io/en/stable/interactive/magics.html) commands. The first command **%%bash** interprets subsequent commands as shell commands

```
%%bash

ls -l
pwd
```

You can get the complete list of magic function commands using %quickref.  
The Quick Reference Card appears as a pane on the right.

```
%quickref
```

Code written in C/C++ can be written and compiled. C++ code is first edited and saved to a file using ***%%writefile*** cell magics command. Subsequently the saved (Hello.cpp) file is compiled and executed the ***%%bash*** cell magics command.

```
%%writefile Hello.cpp

#include <iostream>

int main()
{
    std::cout << "Hello exciting world of Edge AI!" << std::endl;
    return 0;
}

```

Colab is a Linux based machine with C compiler support. C++ code is compiled and executed using bash shell

```
%%bash

g++ -o hello Hello.cpp
./hello
```

- *After executing the C++ code, convert it to C and execute it using **gcc** compiler.*

A Bash command can also be executed by appending ***!*** symbol.  Here we use ***wget*** to upload Pima diabetes dataset.  

```
!wget -q https://edge-ai-doulos.s3.us-west-2.amazonaws.com/diabetes-pima.csv
```
