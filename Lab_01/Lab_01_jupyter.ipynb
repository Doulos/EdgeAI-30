##### **Step 2.** Use of Cell Magics Commands
Colab cells can execute bash shell commands using [Cell Magics](https://ipython.readthedocs.io/en/stable/interactive/magics.html) commands. The first command **%%bash** interprets subsequent commands as shell commands

```python
%%bash

ls -l
pwd
```

##### Saving files using Jupyter cell
Code written in C/C++ can be written and compiled. C++ code is first edited and saved to a file using ***%%writefile*** cell magics command. Subsequently the saved (Hello.cpp) file is compiled and executed the ***%%bash*** cell magics command.

```python
%%writefile Hello.cpp

#include <iostream>

int main()
{
    std::cout << "Hello exciting world of Edge AI!" << std::endl;
    return 0;
}

```

##### Compiling C++ code using GCC
This notebook is hosted on a Linux based machine with C compiler support. C++ code is compiled and executed using bash shell

```python
%%bash

g++ -o hello Hello.cpp
./hello
```

##### Using ! for single line bash commands
