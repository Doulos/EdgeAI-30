### Syntax for explicit control over including header files and linking libraries

Compile source file into an object file, specifying include path
- g++ -c main.cpp -I/path/to/my/headers -o main.o

Link the object file with a shared library, specifying library path
- g++ main.o -L/path/to/my/libraries -lmylib -o my_executable


```python
%%bash

g++ -c opencv-test.cpp -I/usr/include/opencv4 -o opencv-test.o

g++ opencv-test.o -L/usr/lib/x86_64-linux-gnu -lopencv_core -lopencv_highgui -lopencv_imgcodecs -lopencv_videoio -lopencv_imgproc -o opencv-test

./opencv-test
```

Compile application using pkg-config. ***pkg-config*** is used to query information about installed libraries, such as their include paths (where header files are located) and link flags (how to link against the library). Execute the application, which prints out OpenCV build details.

```python
%%bash

g++ -o opencv-test-pkg opencv-test.cpp $(pkg-config --cflags --libs opencv4)

./opencv-test-pkg
```

##### **Step 3(b)**: 

Create CMakeLists.txt file. This CMakeLists.txt file provides instructions to CMake on how to find the OpenCV library and then how to compile and link the opencv-test.cpp file to create an executable program.

```python
%%writefile CMakeLists.txt

cmake_minimum_required(VERSION 3.0)
project(opencv-test)

set(CMAKE_CXX_STANDARD 11)
set(CMAKE_CXX_STANDARD_REQUIRED ON)

find_package(OpenCV REQUIRED)

include_directories(${OpenCV_INCLUDE_DIRS})
add_executable(opencv-test-cmake opencv-test.cpp)
target_link_libraries(opencv-test-cmake ${OpenCV_LIBS})
 
```

Generate binary executable using ***make*** and then execute the application.

##### **Step 4(b)**:  

From the extracted files, copy headers (.h) files to directory */usr/local/include* and the shared object file (.so.1.21.0) to directory */usr/local/lib*. Also create a softlink with .so extension to library file with .so.1.21.0 extension.

```python
import os
import shutil

# Source directory
source_dir = 'onnx_linux_x64_1_21/'

# Writable destination directories in your home folder
include_dir = os.path.expanduser('~/local/include')
lib_dir = os.path.expanduser('~/local/lib')

os.makedirs(include_dir, exist_ok=True)
os.makedirs(lib_dir, exist_ok=True)

# Copy .h files
for root, _, files in os.walk(source_dir):
    for file in files:
        if file.endswith('.h'):
            source_path = os.path.join(root, file)
            destination_path = os.path.join(include_dir, file)
            shutil.copy2(source_path, destination_path)
            print(f"Copied: {source_path} to {destination_path}")

# Copy .so.1.21.0 file and create a symlink
for root, _, files in os.walk(source_dir):
    for file in files:
        if file.endswith('.so.1.21.0'):
            source_path = os.path.join(root, file)
            destination_path = os.path.join(lib_dir, file)
            shutil.copy2(source_path, destination_path)
            print(f"Copied: {source_path} to {destination_path}")

            link_name = file.replace(".1.21.0", "")
            link_path = os.path.join(lib_dir, link_name)

            if os.path.islink(link_path) or os.path.exists(link_path):
                os.remove(link_path)

            os.symlink(destination_path, link_path)
            print(f"Created softlink: {link_path} -> {destination_path}")

print("Copying and linking complete.")
```

##### **Step 4(d)**: 

Create C++ source file that prints out the version of ONNX Runtime and underlying execution provider.

```python
%%writefile onnx-runtime-ver.cpp

#include <iostream>
#include <onnxruntime_cxx_api.h>

int main()
{
    std::string version = Ort::GetVersionString();
    std::cout << version << std::endl;

    std::vector<std::string> providers = Ort::GetAvailableProviders();

    for (const auto &provider: providers) {
        std::cout << provider << std::endl;
    }

    return 0;
}
```
